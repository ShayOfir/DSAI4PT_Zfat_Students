```r
t0 <- Sys.time()
cat("R:", R.version.string, "
")
install.packages("ottr", quiet = TRUE)
cat("install seconds:", round(as.numeric(difftime(Sys.time(), t0, units = "secs")), 1), "
")
cat("ottr version:", as.character(packageVersion("ottr")), "
")
cat("deps:", paste(rownames(installed.packages())[rownames(installed.packages()) %in% c("testthat","R6","jsonlite","rlang")], collapse = ", "), "
")
```

```r
# create a tiny visible test file exactly like otter assign generates
dir.create("tests", showWarnings = FALSE)
writeLines(c(
  "test = list(name = \"q1\", cases = list(",
  "  ottr::TestCase$new(hidden = FALSE, name = NA, points = 1, code = {",
  "    testthat::expect_equal(length(fma), 5)",
  "    testthat::expect_equal(fma_mean, 41.6)",
  "  })))"
), "tests/q1.R")
```

```r
fma <- c(20, 35, 42, 50, 61)
fma_mean <- mean(fma)
t1 <- Sys.time()
. = ottr::check("tests/q1.R")   # expect: All tests passed!
cat("check seconds:", round(as.numeric(difftime(Sys.time(), t1, units = "secs")), 2), "
")
```

```r
# failing case: expect a readable failure message
fma <- c(20, 35, 42, 50)
. = ottr::check("tests/q1.R")
```
