<p dir="rtl" style="direction:rtl; text-align:right;">מחברת חד-פעמית: מודדת התקנת חבילות בסביבת <span dir="ltr">R</span> חדשה ב-<span dir="ltr">Colab</span>, ובודקת שעברית מוצגת נכון.</p>

```r
# Time the installation of each package set on a FRESH Colab R runtime
ua <- sprintf("R/%s R (%s)", getRversion(),
              paste(getRversion(), R.version$platform, R.version$arch, R.version$os))
sets <- c("tidyverse", "tidymodels", "lme4", "glmmTMB", "ggdag")
timeit <- function(repo, pkg) {
  options(repos = c(CRAN = repo), HTTPUserAgent = ua)
  lib <- tempfile("lib"); dir.create(lib)          # empty library: nothing is skipped
  system.time(try(install.packages(pkg, lib = lib, quiet = TRUE)))[["elapsed"]]
}
ppm  <- "https://packagemanager.posit.co/cran/__linux__/jammy/latest"
res  <- data.frame(package = sets, ppm_binary_sec = NA_real_, cran_default_sec = NA_real_)
for (i in seq_along(sets)) res$ppm_binary_sec[i] <- timeit(ppm, sets[i])
# Default CRAN builds from source and can take 10+ minutes per package; set TRUE to measure
RUN_CRAN <- FALSE
if (RUN_CRAN) for (i in seq_along(sets)) res$cran_default_sec[i] <- timeit("https://cloud.r-project.org", sets[i])
print(res)
cat("Total binary-repo seconds:", round(sum(res$ppm_binary_sec)), "\n")
```
