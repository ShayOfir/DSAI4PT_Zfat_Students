```r
library(dplyr)    # load the package (already installed by the setup cell)
higher_fma <- filter(patients, fma_ue >= ___)   # rows that meet the condition
mean_age <- higher_fma |> summarise(m = mean(___)) |> pull(m)   # mean age, as a number
cat("Patients kept:", nrow(higher_fma), "| mean age:", mean_age, "\n")
```

```r
# Visible check: stops with an error if the answer is wrong
stopifnot(nrow(higher_fma) == 3, abs(mean_age - mean(c(71, 64, 49))) < 1e-9)
cat("OK\n")
```
