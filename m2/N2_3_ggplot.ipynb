```r
source("https://raw.githubusercontent.com/ShayOfir/DSAI4PT_Zfat_Students/main/R/load_course_data.R")
stroke <- load_course_data("stroke_clean") |>        # one row per patient
  mutate(lesion_type = factor(lesion_type, levels = c("ischemic", "hemorrhagic", "mixed")),
         age_group = cut(age, c(0, 54, 64, 100),     # three age groups
                         labels = c("under 55", "55-64", "65 and over")))
visits <- load_course_data("stroke_visits") |>       # one row per patient per stage
  mutate(stage = factor(stage, levels = c("subacute", "chronic"))) |>
  left_join(select(stroke, id, lesion_type), by = "id")   # add the stroke type to every visit
glimpse(stroke)
```
